```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime
import re

# Load key datasets
patients_df = pd.read_csv('synthea_sample_data_csv_nov2021/csv/patients.csv')
conditions_df = pd.read_csv('synthea_sample_data_csv_nov2021/csv/conditions.csv')
encounters_df = pd.read_csv('synthea_sample_data_csv_nov2021/csv/encounters.csv')
medications_df = pd.read_csv('synthea_sample_data_csv_nov2021/csv/medications.csv')

# Display basic patient information
print(f"Total patients: {patients_df.shape[0]}")
print(f"Patient columns: {patients_df.columns.tolist()}")

# Calculate age
# Convert birthdate to datetime
patients_df['BIRTHDATE'] = pd.to_datetime(patients_df['BIRTHDATE'])
# Calculate age based on today's date
current_date = datetime.now()
patients_df['AGE'] = ((current_date - patients_df['BIRTHDATE']).dt.days / 365.25).round().astype(int)

# Identify target groups
# 1. Minors (under 18)
minors = patients_df[patients_df['AGE'] < 18]

# 2. Elderly (65 and older)
elderly = patients_df[patients_df['AGE'] >= 65]

# 3. Middle-aged adults (18-64)
adults = patients_df[(patients_df['AGE'] >= 18) & (patients_df['AGE'] < 65)]

print(f"\nAge groups:")
print(f"Minors (<18): {len(minors)} ({len(minors)/len(patients_df)*100:.1f}%)")
print(f"Adults (18-64): {len(adults)} ({len(adults)/len(patients_df)*100:.1f}%)")
print(f"Elderly (65+): {len(elderly)} ({len(elderly)/len(patients_df)*100:.1f}%)")

# Plot age distribution
plt.figure(figsize=(10, 6))
sns.histplot(patients_df['AGE'], bins=20)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.axvline(x=18, color='r', linestyle='--', label='Minor cutoff (18)')
plt.axvline(x=65, color='g', linestyle='--', label='Elderly cutoff (65)')
plt.legend()
plt.show()

# Check race/ethnicity distribution (potential language proxies)
print("\nRace distribution:")
print(patients_df['RACE'].value_counts())

print("\nEthnicity distribution:")
print(patients_df['ETHNICITY'].value_counts())
```

```python
# Identify patients with special communication needs based on conditions
# Convert descriptions to lowercase for case-insensitive search
conditions_df['DESCRIPTION_lower'] = conditions_df['DESCRIPTION'].str.lower()

# Define conditions that might require special communication
special_needs_conditions = [
    'alzheimer', 'dementia', 'autism', 'autistic', 'asperger', 
    'intellectual disability', 'brain injury', 'tbi', 'stroke', 
    'aphasia', 'schizophrenia', 'bipolar', 'learning disorder', 
    'mental disorder', 'anxiety', 'depression'
]

# Create dictionary to store patients with specific conditions
condition_patients = {}

for condition in special_needs_conditions:
    # Find patients with this condition
    matching = conditions_df[conditions_df['DESCRIPTION_lower'].str.contains(condition, na=False)]
    patient_ids = matching['PATIENT'].unique()
    
    if len(patient_ids) > 0:
        condition_patients[condition] = patient_ids
        print(f"Found {len(patient_ids)} patients with '{condition}' related conditions")
        
        # Show a few example descriptions
        examples = matching['DESCRIPTION'].unique()[:3]
        print(f"Example conditions: {examples}")
        print()

# Create a set of all patients with special communication needs
all_special_needs_patients = set()
for patients in condition_patients.values():
    all_special_needs_patients.update(patients)

print(f"\nTotal unique patients with special communication needs: {len(all_special_needs_patients)}")

# Join with patient data to see demographics of special needs patients
special_patients_df = patients_df[patients_df['Id'].isin(all_special_needs_patients)]

print(f"\nAge distribution of patients with special communication needs:")
print(f"Minors: {sum(special_patients_df['AGE'] < 18)}")
print(f"Adults: {sum((special_patients_df['AGE'] >= 18) & (special_patients_df['AGE'] < 65))}")
print(f"Elderly: {sum(special_patients_df['AGE'] >= 65)}")
```

```python
# Examine encounter data for notes
print("\nEncounter data:")
print(f"Total encounters: {encounters_df.shape[0]}")
print(f"Encounters with reason descriptions: {encounters_df['REASONDESCRIPTION'].notna().sum()}")

# Show some example reason descriptions
print("\nSample reason descriptions:")
sample_descriptions = encounters_df[encounters_df['REASONDESCRIPTION'].notna()]['REASONDESCRIPTION'].sample(min(5, encounters_df['REASONDESCRIPTION'].notna().sum()))
for desc in sample_descriptions:
    print(f"- {desc}")

# Let's check if we can find discharge summaries
discharge_encounters = encounters_df[
    (encounters_df['DESCRIPTION'].str.contains('discharge', case=False, na=False)) |
    (encounters_df['REASONDESCRIPTION'].str.contains('discharge', case=False, na=False))
]

print(f"\nFound {len(discharge_encounters)} encounters related to discharge")

# Check medication data for our target groups
print("\nMedication data:")
print(f"Total medication records: {medications_df.shape[0]}")

# Get a sense of what diagnoses require medications
med_reasons = medications_df[medications_df['REASONDESCRIPTION'].notna()]['REASONDESCRIPTION'].value_counts().head(10)
print("\nTop 10 reasons for medications:")
print(med_reasons)
```

```python
def get_patient_summary(patient_id):
    """Create a comprehensive medical summary for a patient"""
    
    # Get patient demographics
    patient = patients_df[patients_df['Id'] == patient_id].iloc[0]
    
    # Get patient conditions
    patient_conditions = conditions_df[conditions_df['PATIENT'] == patient_id]
    current_conditions = patient_conditions[patient_conditions['STOP'].isna()]
    
    # Get medications
    patient_meds = medications_df[medications_df['PATIENT'] == patient_id]
    current_meds = patient_meds[patient_meds['STOP'].isna()]
    
    # Get recent encounters
    patient_encounters = encounters_df[encounters_df['PATIENT'] == patient_id]
    recent_encounters = patient_encounters.sort_values('START', ascending=False).head(5)
    
    # Create summary dictionary
    summary = {
        'patient_info': {
            'id': patient_id,
            'age': patient['AGE'],
            'gender': patient['GENDER'],
            'race': patient['RACE'],
            'ethnicity': patient['ETHNICITY'],
        },
        'current_conditions': current_conditions['DESCRIPTION'].tolist() if not current_conditions.empty else [],
        'all_conditions': patient_conditions['DESCRIPTION'].tolist() if not patient_conditions.empty else [],
        'current_medications': current_meds['DESCRIPTION'].tolist() if not current_meds.empty else [],
        'recent_encounters': []
    }
    
    # Add encounter details
    for _, enc in recent_encounters.iterrows():
        encounter_info = {
            'date': enc['START'],
            'type': enc['ENCOUNTERCLASS'],
            'description': enc['DESCRIPTION'],
            'reason': enc['REASONDESCRIPTION'] if pd.notna(enc['REASONDESCRIPTION']) else 'No reason provided'
        }
        summary['recent_encounters'].append(encounter_info)
    
    return summary

# Test the function with a random patient
test_patient = patients_df['Id'].sample(1).iloc[0]
patient_summary = get_patient_summary(test_patient)

# Print the summary
print(f"\nSample patient summary for patient {test_patient}:")
print(f"Age: {patient_summary['patient_info']['age']}, Gender: {patient_summary['patient_info']['gender']}")
print(f"Current conditions: {patient_summary['current_conditions'][:3]}...")
print(f"Current medications: {patient_summary['current_medications'][:3]}...")
print(f"Recent encounters: {len(patient_summary['recent_encounters'])}")
```

```python
import requests
def get_complete_patient_history(patient_id):
    """
    Create a comprehensive medical history for a patient including all relevant
    information from encounters, conditions, medications, procedures, etc.
    """
    # Get patient demographics
    patient = patients_df[patients_df['Id'] == patient_id].iloc[0]
    
    # Get all patient data from different tables
    patient_conditions = conditions_df[conditions_df['PATIENT'] == patient_id].sort_values('START')
    patient_medications = medications_df[medications_df['PATIENT'] == patient_id].sort_values('START')
    patient_encounters = encounters_df[encounters_df['PATIENT'] == patient_id].sort_values('START')
    
    # Try to get additional data if available
    try:
        patient_procedures = pd.read_csv('synthea_sample_data_csv_nov2021/csv/procedures.csv')
        patient_procedures = patient_procedures[patient_procedures['PATIENT'] == patient_id].sort_values('START')
    except:
        patient_procedures = pd.DataFrame(columns=['START', 'DESCRIPTION'])
    
    try:
        patient_allergies = pd.read_csv('synthea_sample_data_csv_nov2021/csv/allergies.csv')
        patient_allergies = patient_allergies[patient_allergies['PATIENT'] == patient_id]
    except:
        patient_allergies = pd.DataFrame(columns=['DESCRIPTION'])
    
    # Build patient profile
    profile = {
        'id': patient_id,
        'basic_info': {
            'age': patient['AGE'],
            'gender': patient['GENDER'],
            'race': patient['RACE'],
            'ethnicity': patient['ETHNICITY'],
            'birthdate': patient['BIRTHDATE'],
            'marital_status': patient['MARITAL'] if 'MARITAL' in patient else 'Unknown',
            'address': f"{patient['ADDRESS']}, {patient['CITY']}, {patient['STATE']} {patient['ZIP']}",
        },
        'conditions': [],
        'medications': [],
        'encounters': [],
        'procedures': [],
        'allergies': patient_allergies['DESCRIPTION'].tolist() if not patient_allergies.empty else []
    }
    
    # Process conditions
    for _, condition in patient_conditions.iterrows():
        condition_info = {
            'date': condition['START'],
            'end_date': condition['STOP'] if pd.notna(condition['STOP']) else 'Ongoing',
            'description': condition['DESCRIPTION'],
            'code': condition['CODE']
        }
        profile['conditions'].append(condition_info)
    
    # Process medications
    for _, med in patient_medications.iterrows():
        med_info = {
            'date': med['START'],
            'end_date': med['STOP'] if pd.notna(med['STOP']) else 'Ongoing',
            'description': med['DESCRIPTION'],
            'reason': med['REASONDESCRIPTION'] if pd.notna(med['REASONDESCRIPTION']) else 'Not specified'
        }
        profile['medications'].append(med_info)
    
    # Process encounters
    for _, enc in patient_encounters.iterrows():
        enc_info = {
            'date': enc['START'],
            'end_date': enc['STOP'],
            'type': enc['ENCOUNTERCLASS'],
            'description': enc['DESCRIPTION'],
            'reason': enc['REASONDESCRIPTION'] if pd.notna(enc['REASONDESCRIPTION']) else 'Not specified',
            'provider': enc['PROVIDER'],
            'organization': enc['ORGANIZATION']
        }
        profile['encounters'].append(enc_info)
    
    # Process procedures
    for _, proc in patient_procedures.iterrows():
        proc_info = {
            'date': proc['START'],
            'description': proc['DESCRIPTION'],
            'reason': proc['REASONDESCRIPTION'] if pd.notna(proc['REASONDESCRIPTION']) else 'Not specified'
        }
        profile['procedures'].append(proc_info)
    
    return profile

def format_medical_history_prompt(patient_history, target_audience):
    """
    Format the patient history into a prompt for the LLM to generate
    a simplified discharge summary for the specified target audience.
    """
    # Define prompt based on target audience
    if target_audience == "pediatric":
        audience_instruction = "The patient is a minor (under 18). Create a simplified discharge summary that would be appropriate for a child or adolescent patient and their parents/guardians."
    elif target_audience == "elderly":
        audience_instruction = "The patient is elderly (65+). Create a simplified discharge summary with larger text indicators [LARGE], clear medication instructions, and straightforward follow-up directions."
    elif target_audience == "cognitive_impairment":
        audience_instruction = "The patient has a condition that may affect cognition (e.g., dementia, intellectual disability, brain injury). Create a very simplified discharge summary with visual cues, simple language, and step-by-step instructions."
    elif target_audience == "esl":
        audience_instruction = "Create a discharge summary for someone with limited English proficiency. Use simple vocabulary, avoid idioms, and include visual indicators where helpful."
    else:
        audience_instruction = "Create a standard discharge summary in simple language that most patients can understand."
    
    # Format patient demographics
    basic_info = patient_history['basic_info']
    demographics = f"""
Patient Demographics:
- Age: {basic_info['age']}
- Gender: {basic_info['gender']}
- Race: {basic_info['race']}
- Ethnicity: {basic_info['ethnicity']}
"""
    
    # Format current conditions
    current_conditions = [c for c in patient_history['conditions'] if c['end_date'] == 'Ongoing']
    conditions_text = "Current Conditions:\n"
    for cond in current_conditions:
        conditions_text += f"- {cond['description']} (since {cond['date']})\n"
    
    # Format current medications
    current_medications = [m for m in patient_history['medications'] if m['end_date'] == 'Ongoing']
    medications_text = "Current Medications:\n"
    for med in current_medications:
        medications_text += f"- {med['description']} (Reason: {med['reason']})\n"
    
    # Format recent encounters
    sorted_encounters = sorted(patient_history['encounters'], key=lambda x: x['date'], reverse=True)
    recent_encounters = sorted_encounters[:5] if len(sorted_encounters) > 5 else sorted_encounters
    encounters_text = "Recent Medical Visits:\n"
    for enc in recent_encounters:
        encounters_text += f"- {enc['date']}: {enc['description']} (Reason: {enc['reason']})\n"
    
    # Format recent procedures
    sorted_procedures = sorted(patient_history['procedures'], key=lambda x: x['date'], reverse=True)
    recent_procedures = sorted_procedures[:5] if len(sorted_procedures) > 5 else sorted_procedures
    procedures_text = "Recent Procedures:\n"
    for proc in recent_procedures:
        procedures_text += f"- {proc['date']}: {proc['description']} (Reason: {proc['reason']})\n"
    
    # Format allergies
    allergies_text = "Allergies:\n"
    if patient_history['allergies']:
        for allergy in patient_history['allergies']:
            allergies_text += f"- {allergy}\n"
    else:
        allergies_text += "- No known allergies\n"
    
    # Create the full prompt
    prompt = f"""
You are a medical communication specialist who creates simplified discharge summaries for patients with different needs. Your task is to create a clear, understandable discharge summary based on the following patient information.

{audience_instruction}

PATIENT MEDICAL HISTORY:
{demographics}
{conditions_text}
{medications_text}
{encounters_text}
{procedures_text}
{allergies_text}

Based on this information, create a simplified discharge summary that:
1. Summarizes the patient's condition in plain language
2. Explains their medications in an easy-to-understand way
3. Provides clear follow-up instructions
4. Includes warning signs that should prompt them to seek immediate care
5. Uses appropriate visual formatting for the target audience

The summary should be comprehensive but easy to understand for the specified target audience.
"""
    
    return prompt

def generate_simplified_discharge_summary(patient_id, target_audience="standard"):
    """
    Generate a simplified discharge summary for a patient using the LLM.
    """
    # Get patient history
    patient_history = get_complete_patient_history(patient_id)
    
    # Format the prompt
    prompt = format_medical_history_prompt(patient_history, target_audience)
    
    # Print the prompt for debugging
    print(f"\nGenerated prompt for patient {patient_id}:\n{prompt}")
    print(f"Target audience: {target_audience}")
    print(f"Patient ID: {patient_id}")
    print(f"Patient history: {patient_history}")
    print(f"Prompt length: {len(prompt)} characters")
    print(f"Prompt length: {len(prompt.split())} words")
    print(f"Prompt length: {len(re.findall(r'\w+', prompt))} tokens")

    # Measure the tokens of the prompt
    try:
        import tiktoken
        encoding = tiktoken.get_encoding("cl100k_base")
        num_tokens = len(encoding.encode(prompt))
        print(f"Number of tokens in the prompt: {num_tokens}")
    except:
        print("Couldn't measure token count - tiktoken may not be installed")
    
    # Send the prompt to the LLM
    try:
        response = requests.post(
            "http://10.20.31.1:11434/api/chat",
            json={
                "model": "aravhawk/llama4",
                "messages": [{"role": "user", "content": prompt}],
                "stream": False
            }
        )
        
        # Get the response content
        response_data = response.json()
        simplified_summary = response_data["message"]["content"]
        
        return {
            "patient_id": patient_id,
            "target_audience": target_audience,
            "patient_history": patient_history,
            "simplified_summary": simplified_summary
        }
    except Exception as e:
        print(f"Error generating summary: {str(e)}")
        return None

# Test the function with a random patient from each target group
import random

# Define our target groups
print("Finding sample patients for each target group...")

# 1. Pediatric patient
random_pediatric = minors['Id'].sample(1).iloc[0]
print(f"Selected pediatric patient ID: {random_pediatric}")

# 2. Elderly patient
random_elderly = elderly['Id'].sample(1).iloc[0]
print(f"Selected elderly patient ID: {random_elderly}")

# 3. Patient with potential cognitive impairment
if len(all_special_needs_patients) > 0:
    random_special_needs = random.choice(list(all_special_needs_patients))
    print(f"Selected special needs patient ID: {random_special_needs}")
else:
    random_special_needs = None
    print("No patients with cognitive impairment found")

# Test with one patient
test_patient_id = random_elderly  # Change this to test different patient types
test_audience = "elderly"  # Options: "pediatric", "elderly", "cognitive_impairment", "esl", "standard"
print(f"\nGenerating discharge summary for {test_audience} patient {test_patient_id}...")

result = generate_simplified_discharge_summary(test_patient_id, test_audience)

if result:
    print("\nSimplified Discharge Summary:")
    from IPython.display import Markdown
    display(Markdown(result["simplified_summary"]))
```

### Summaries for patients with multiple types of impairments (child, elderly, cognitively-impaired)

```python
import requests
import pandas as pd
import re
import random
from IPython.display import Markdown

def get_complete_patient_history(patient_id):
    """
    Create a comprehensive medical history for a patient including all relevant
    information from encounters, conditions, medications, procedures, etc.
    """
    # Get patient demographics - include ALL available fields
    patient = patients_df[patients_df['Id'] == patient_id].iloc[0]
    
    # Get all patient data from different tables
    patient_conditions = conditions_df[conditions_df['PATIENT'] == patient_id].sort_values('START')
    patient_medications = medications_df[medications_df['PATIENT'] == patient_id].sort_values('START')
    patient_encounters = encounters_df[encounters_df['PATIENT'] == patient_id].sort_values('START')
    
    # Try to get additional data if available
    try:
        patient_procedures = pd.read_csv('synthea_sample_data_csv_nov2021/csv/procedures.csv')
        patient_procedures = patient_procedures[patient_procedures['PATIENT'] == patient_id].sort_values('START')
    except:
        patient_procedures = pd.DataFrame(columns=['START', 'DESCRIPTION'])
    
    try:
        patient_allergies = pd.read_csv('synthea_sample_data_csv_nov2021/csv/allergies.csv')
        patient_allergies = patient_allergies[patient_allergies['PATIENT'] == patient_id]
    except:
        patient_allergies = pd.DataFrame(columns=['DESCRIPTION'])
    
    # Try to get social determinants of health if available
    try:
        patient_social = pd.read_csv('synthea_sample_data_csv_nov2021/csv/social_determinants.csv')
        patient_social = patient_social[patient_social['PATIENT'] == patient_id]
    except:
        patient_social = pd.DataFrame(columns=['CATEGORY', 'DESCRIPTION'])
    
    # Get preferred name if available (could be in observations or patient_attributes table)
    preferred_name = None
    try:
        observations = pd.read_csv('synthea_sample_data_csv_nov2021/csv/observations.csv')
        name_obs = observations[(observations['PATIENT'] == patient_id) & 
                               (observations['CODE'].isin(['preferred_name', 'PREF_NAME', 'preferred name']))]
        if not name_obs.empty:
            preferred_name = name_obs.iloc[0]['VALUE']
    except:
        pass
    
    # Calculate age from birthdate
    from datetime import datetime
    birthdate = pd.to_datetime(patient['BIRTHDATE'])
    age = (datetime.now() - birthdate).days // 365
    
    # Build enhanced patient profile
    profile = {
        'id': patient_id,
        'basic_info': {
            'first_name': patient['FIRST'],
            'last_name': patient['LAST'],
            'preferred_name': preferred_name if preferred_name else patient['FIRST'],
            'prefix': patient['PREFIX'] if 'PREFIX' in patient and pd.notna(patient['PREFIX']) else None,
            'suffix': patient['SUFFIX'] if 'SUFFIX' in patient and pd.notna(patient['SUFFIX']) else None,
            'age': age,
            'birthdate': patient['BIRTHDATE'],
            'gender': patient['GENDER'],
            'race': patient['RACE'],
            'ethnicity': patient['ETHNICITY'],
            'marital_status': patient['MARITAL'] if 'MARITAL' in patient and pd.notna(patient['MARITAL']) else 'Unknown',
            'birthplace': patient['BIRTHPLACE'] if 'BIRTHPLACE' in patient and pd.notna(patient['BIRTHPLACE']) else 'Unknown',
            'address': f"{patient['ADDRESS']}, {patient['CITY']}, {patient['STATE']} {patient['ZIP']}",
            'healthcare_expenses': patient['HEALTHCARE_EXPENSES'] if 'HEALTHCARE_EXPENSES' in patient else None,
            'healthcare_coverage': patient['HEALTHCARE_COVERAGE'] if 'HEALTHCARE_COVERAGE' in patient else None,
        },
        'conditions': [],
        'medications': [],
        'encounters': [],
        'procedures': [],
        'allergies': patient_allergies['DESCRIPTION'].tolist() if not patient_allergies.empty else [],
        'social_determinants': []
    }
    
    # Process social determinants
    for _, social in patient_social.iterrows():
        social_info = {
            'category': social['CATEGORY'] if 'CATEGORY' in social else 'Unknown',
            'description': social['DESCRIPTION'] if 'DESCRIPTION' in social else 'Unknown'
        }
        profile['social_determinants'].append(social_info)
    
    # Process conditions
    for _, condition in patient_conditions.iterrows():
        condition_info = {
            'date': condition['START'],
            'end_date': condition['STOP'] if pd.notna(condition['STOP']) else 'Ongoing',
            'description': condition['DESCRIPTION'],
            'code': condition['CODE']
        }
        profile['conditions'].append(condition_info)
    
    # Process medications
    for _, med in patient_medications.iterrows():
        med_info = {
            'date': med['START'],
            'end_date': med['STOP'] if pd.notna(med['STOP']) else 'Ongoing',
            'description': med['DESCRIPTION'],
            'reason': med['REASONDESCRIPTION'] if pd.notna(med['REASONDESCRIPTION']) else 'Not specified',
            'dispense': med['DISPENSES'] if 'DISPENSES' in med and pd.notna(med['DISPENSES']) else None,
            'dosage': med['DOSAGE'] if 'DOSAGE' in med and pd.notna(med['DOSAGE']) else None,
            'instructions': med['INSTRUCTIONS'] if 'INSTRUCTIONS' in med and pd.notna(med['INSTRUCTIONS']) else None
        }
        profile['medications'].append(med_info)
    
    # Process encounters
    for _, enc in patient_encounters.iterrows():
        enc_info = {
            'date': enc['START'],
            'end_date': enc['STOP'],
            'type': enc['ENCOUNTERCLASS'],
            'description': enc['DESCRIPTION'],
            'reason': enc['REASONDESCRIPTION'] if pd.notna(enc['REASONDESCRIPTION']) else 'Not specified',
            'provider': enc['PROVIDER'],
            'organization': enc['ORGANIZATION']
        }
        profile['encounters'].append(enc_info)
        
        # Find most recent primary care provider
        if enc['ENCOUNTERCLASS'] == 'ambulatory' and 'primary care' in enc['DESCRIPTION'].lower():
            profile['basic_info']['primary_provider'] = enc['PROVIDER']
            profile['basic_info']['primary_organization'] = enc['ORGANIZATION']
    
    # Process procedures
    for _, proc in patient_procedures.iterrows():
        proc_info = {
            'date': proc['START'],
            'description': proc['DESCRIPTION'],
            'reason': proc['REASONDESCRIPTION'] if pd.notna(proc['REASONDESCRIPTION']) else 'Not specified'
        }
        profile['procedures'].append(proc_info)
    
    return profile

def format_medical_history_prompt(patient_history, target_audience):
    """
    Format the patient history into a personalized prompt for the LLM to generate
    a simplified discharge summary for the specified target audience.
    """
    # Define prompt based on target audience
    if target_audience == "pediatric":
        audience_instruction = f"""The patient is a minor (age {patient_history['basic_info']['age']}). 
Create a simplified discharge summary that would be appropriate for a child or adolescent patient and their parents/guardians.
Use warm, reassuring language and simple explanations of medical concepts. 
If the child is very young (under 10), include elements that make the summary more engaging for kids.
Address both the child and the parent/guardian in the summary."""
    elif target_audience == "elderly":
        audience_instruction = f"""The patient is elderly (age {patient_history['basic_info']['age']}). 
Create a simplified discharge summary with visual indicators for important sections, clear medication instructions, 
and straightforward follow-up directions. Be respectful and avoid condescending language.
Remember that elderly patients often have more complex health situations and may be taking multiple medications."""
    elif target_audience == "cognitive_impairment":
        audience_instruction = """The patient has a condition that may affect cognition (e.g., dementia, intellectual disability, brain injury). 
Create a very simplified discharge summary with visual cues, simple language, step-by-step instructions, and information directed at both the patient and their caregiver(s).
Use short sentences, simple words, and clear organization."""
    elif target_audience == "esl":
        audience_instruction = """Create a discharge summary for someone with limited English proficiency. 
Use simple vocabulary, avoid idioms and medical jargon, and include visual indicators where helpful.
Use short, clear sentences and provide explanations for necessary medical terms."""
    else:
        audience_instruction = """Create a standard discharge summary in simple language that most patients can understand.
Balance medical accuracy with clear explanations."""
    
    # Format patient demographics with more personalized information
    basic_info = patient_history['basic_info']
    
    # Build formal name with prefix/suffix if available
    full_name = ""
    if basic_info['prefix']:
        full_name += f"{basic_info['prefix']} "
    full_name += f"{basic_info['first_name']} {basic_info['last_name']}"
    if basic_info['suffix']:
        full_name += f" {basic_info['suffix']}"
    
    preferred_name = basic_info['preferred_name'] if basic_info['preferred_name'] else basic_info['first_name']
    
    demographics = f"""
Patient Information:
- Full Name: {full_name}
- Preferred Name: {preferred_name}
- Age: {basic_info['age']}
- Gender: {basic_info['gender']}
- Race: {basic_info['race']}
- Ethnicity: {basic_info['ethnicity']}
- Marital Status: {basic_info['marital_status']}
- Birthplace: {basic_info['birthplace']}
- Address: {basic_info['address']}
"""
    
    # Add primary care provider info if available
    if 'primary_provider' in basic_info and basic_info['primary_provider']:
        demographics += f"- Primary Care Provider: {basic_info['primary_provider']}\n"
        demographics += f"- Primary Care Organization: {basic_info['primary_organization']}\n"
    
    # Add social determinants if available
    social_info = ""
    if patient_history['social_determinants']:
        social_info = "\nSocial Context:\n"
        for social in patient_history['social_determinants']:
            social_info += f"- {social['category']}: {social['description']}\n"
    
    # Format current conditions
    current_conditions = [c for c in patient_history['conditions'] if c['end_date'] == 'Ongoing']
    conditions_text = "Current Health Conditions:\n"
    for cond in current_conditions:
        conditions_text += f"- {cond['description']} (since {cond['date']})\n"
    
    # Format current medications with more details
    current_medications = [m for m in patient_history['medications'] if m['end_date'] == 'Ongoing']
    medications_text = "Current Medications:\n"
    for med in current_medications:
        med_text = f"- {med['description']} (Reason: {med['reason']})"
        if med['dosage']:
            med_text += f", Dosage: {med['dosage']}"
        if med['instructions']:
            med_text += f", Instructions: {med['instructions']}"
        medications_text += med_text + "\n"
    
    # Format recent encounters
    sorted_encounters = sorted(patient_history['encounters'], key=lambda x: x['date'], reverse=True)
    recent_encounters = sorted_encounters[:5] if len(sorted_encounters) > 5 else sorted_encounters
    encounters_text = "Recent Medical Visits:\n"
    for enc in recent_encounters:
        encounters_text += f"- {enc['date']}: {enc['description']} at {enc['organization']} (Reason: {enc['reason']})\n"
    
    # Format recent procedures
    sorted_procedures = sorted(patient_history['procedures'], key=lambda x: x['date'], reverse=True)
    recent_procedures = sorted_procedures[:5] if len(sorted_procedures) > 5 else sorted_procedures
    procedures_text = "Recent Procedures:\n"
    if recent_procedures:
        for proc in recent_procedures:
            procedures_text += f"- {proc['date']}: {proc['description']} (Reason: {proc['reason']})\n"
    else:
        procedures_text += "- No recent procedures on record\n"
    
    # Format allergies
    allergies_text = "Allergies:\n"
    if patient_history['allergies']:
        for allergy in patient_history['allergies']:
            allergies_text += f"- {allergy}\n"
    else:
        allergies_text += "- No known allergies\n"
    
    # Create the full prompt with more personalization instructions
    prompt = f"""
You are a compassionate medical communication specialist who creates personalized discharge summaries for patients with different needs. Your task is to create a clear, understandable, and personalized discharge summary based on the following patient information.

{audience_instruction}

PATIENT INFORMATION:
{demographics}
{social_info}
{conditions_text}
{medications_text}
{encounters_text}
{procedures_text}
{allergies_text}

Create a personalized discharge summary that:
1. Addresses the patient by their preferred name ({preferred_name}) and makes them feel seen as a person, not just a medical case
2. Summarizes their condition in plain language
3. Explains their medications clearly with a simple schedule
4. Provides specific follow-up instructions
5. Includes warning signs that should prompt them to seek immediate care
6. Uses appropriate visual formatting for the target audience

Write in a warm, empathetic tone that acknowledges the patient's individual circumstances. Avoid generic language. Make the patient feel that this summary was created specifically for them.

Important discharge information should be clearly highlighted. Medication instructions should be particularly clear and easy to follow.
"""
    
    return prompt

def generate_simplified_discharge_summary(patient_id, target_audience="standard"):
    """
    Generate a personalized simplified discharge summary for a patient using the LLM.
    """
    # Get patient history
    patient_history = get_complete_patient_history(patient_id)
    
    # Format the prompt
    prompt = format_medical_history_prompt(patient_history, target_audience)
    
    # Print the prompt for debugging
    print(f"\nGenerated prompt for patient {patient_id}:\n{prompt}")
    print(f"Target audience: {target_audience}")
    
    # Measure the tokens of the prompt
    try:
        import tiktoken
        encoding = tiktoken.get_encoding("cl100k_base")
        num_tokens = len(encoding.encode(prompt))
        print(f"Number of tokens in the prompt: {num_tokens}")
    except:
        print("Couldn't measure token count - tiktoken may not be installed")
    
    # Send the prompt to the LLM
    try:
        response = requests.post(
            "http://10.20.31.1:11434/api/chat",
            json={
                "model": "aravhawk/llama4",
                "messages": [{"role": "user", "content": prompt}],
                "stream": False
            }
        )
        
        # Get the response content
        response_data = response.json()
        simplified_summary = response_data["message"]["content"]
        
        return {
            "patient_id": patient_id,
            "target_audience": target_audience,
            "patient_name": f"{patient_history['basic_info']['first_name']} {patient_history['basic_info']['last_name']}",
            "patient_age": patient_history['basic_info']['age'],
            "simplified_summary": simplified_summary
        }
    except Exception as e:
        print(f"Error generating summary: {str(e)}")
        return None

# Run tests for each patient category
def run_category_tests():
    print("Running tests for each patient category...\n")
    results = []
    
    # 1. Pediatric patient
    random_pediatric = minors['Id'].sample(1).iloc[0]
    print(f"Selected pediatric patient ID: {random_pediatric}")
    pediatric_result = generate_simplified_discharge_summary(random_pediatric, "pediatric")
    if pediatric_result:
        results.append(pediatric_result)
        print(f"\nGenerated pediatric discharge summary for {pediatric_result['patient_name']} (Age: {pediatric_result['patient_age']})")
    
    # 2. Elderly patient
    random_elderly = elderly['Id'].sample(1).iloc[0]
    print(f"\nSelected elderly patient ID: {random_elderly}")
    elderly_result = generate_simplified_discharge_summary(random_elderly, "elderly")
    if elderly_result:
        results.append(elderly_result)
        print(f"\nGenerated elderly discharge summary for {elderly_result['patient_name']} (Age: {elderly_result['patient_age']})")
    
    # 3. Patient with potential cognitive impairment (or another elderly if none found)
    if all_special_needs_patients and len(all_special_needs_patients) > 0:
        random_special_needs = random.choice(list(all_special_needs_patients))
        print(f"\nSelected cognitive impairment patient ID: {random_special_needs}")
        special_needs_result = generate_simplified_discharge_summary(random_special_needs, "cognitive_impairment")
    else:
        print("\nNo patients with cognitive impairment found. Using another elderly patient.")
        random_special_needs = elderly['Id'].sample(1).iloc[0]
        print(f"Selected alternative elderly patient ID: {random_special_needs}")
        special_needs_result = generate_simplified_discharge_summary(random_special_needs, "cognitive_impairment")
    
    if special_needs_result:
        results.append(special_needs_result)
        print(f"\nGenerated cognitive impairment discharge summary for {special_needs_result['patient_name']} (Age: {special_needs_result['patient_age']})")
    
    # Display results
    print("\n========== DISCHARGE SUMMARIES ==========\n")
    for result in results:
        print(f"\n----- {result['target_audience'].upper()} DISCHARGE SUMMARY FOR {result['patient_name']} (Age: {result['patient_age']}) -----\n")
        display(Markdown(result["simplified_summary"]))
        print("\n" + "-"*80 + "\n")
    
    return results

# Run all tests
test_results = run_category_tests()
```

# Final Summmary Generator

### Key Considerations
- Dynamic Prompting based on Patient Age, Patient health Conditions, and Patient Symptoms
- Edge Cases such as Death, Visits with Minors, and Social determinants of health counseling considered
- Pipeline of LLM's to generate summaries

### Technical Details
- Data is extraction and formatting into a round 1 prompt
  - Prompt is generated based on the patient age, health conditions, medications, diagnosis, and symptoms throughout their entire chart history
  - Prompt contains all of the patient history collected, across multiple visits, and inferences times of different procedures, prescriptions, and diagnoses

- Deepseek R1 First Pass Analysis (**Tree-of-Thought**)
  - Utilizes Deepseek R1 for case review and prompt generation, outputting the thinking process and key considerations to consider in the output summary
  - Deepseek generates points of focus and makes reccomendations, as well as thinks through the case in the <think> </think> tags, which are appended to the final prompt

- Few Shot Prompting (**In-Context-Learning**)
  - Data inference is used to determine if the patient is a minior, if the patient is deceased, and if the patient has any social determinants of health
  - A reference template is chosen that matches the type of case (e.g., minor, deceased, etc.) and included in the prompt
  - Deviations from the prompt are noted and included in the final summary

- Final Case Summarization using Llama4
  - The final case summary is generated using llama4, a recently released opensource model
  - The prompt contains all of the patient history and data collected, the results from deepseeks initial analysis, and an example format that was provided via few shot prompting

- The final summary is then formatted and returned to the user in a structured format

```python
import requests
import pandas as pd
import re
import random
from datetime import datetime
from IPython.display import Markdown

def get_complete_patient_history(patient_id):
    """
    Create a comprehensive medical history for a patient including all relevant
    information from encounters, conditions, medications, procedures, etc.
    Focus on the most recent visit and current information.
    """
    # Get patient demographics - include ALL available fields
    patient = patients_df[patients_df['Id'] == patient_id].iloc[0]
    
    # Get all patient data from different tables
    patient_conditions = conditions_df[conditions_df['PATIENT'] == patient_id].sort_values('START')
    patient_medications = medications_df[medications_df['PATIENT'] == patient_id].sort_values('START')
    patient_encounters = encounters_df[encounters_df['PATIENT'] == patient_id].sort_values('START', ascending=False)
    
    # Try to get additional data if available
    try:
        patient_procedures = pd.read_csv('synthea_sample_data_csv_nov2021/csv/procedures.csv')
        patient_procedures = patient_procedures[patient_procedures['PATIENT'] == patient_id].sort_values('START', ascending=False)
    except:
        patient_procedures = pd.DataFrame(columns=['START', 'DESCRIPTION'])
    
    try:
        patient_allergies = pd.read_csv('synthea_sample_data_csv_nov2021/csv/allergies.csv')
        patient_allergies = patient_allergies[patient_allergies['PATIENT'] == patient_id]
    except:
        patient_allergies = pd.DataFrame(columns=['DESCRIPTION'])
    
    # Try to get social determinants of health if available
    try:
        patient_social = pd.read_csv('synthea_sample_data_csv_nov2021/csv/social_determinants.csv')
        patient_social = patient_social[patient_social['PATIENT'] == patient_id]
    except:
        patient_social = pd.DataFrame(columns=['CATEGORY', 'DESCRIPTION'])
    
    # Get preferred name if available (could be in observations or patient_attributes table)
    preferred_name = None
    try:
        observations = pd.read_csv('synthea_sample_data_csv_nov2021/csv/observations.csv')
        name_obs = observations[(observations['PATIENT'] == patient_id) & 
                               (observations['CODE'].isin(['preferred_name', 'PREF_NAME', 'preferred name']))]
        if not name_obs.empty:
            preferred_name = name_obs.iloc[0]['VALUE']
    except:
        pass
    
    # Calculate age from birthdate
    birthdate = pd.to_datetime(patient['BIRTHDATE'])
    age = (datetime.now() - birthdate).days // 365
    
    # Identify most recent encounter (the one they're being discharged from)
    most_recent_encounter = None
    most_recent_encounter_date = None
    most_recent_diagnosis = None
    most_recent_reason = None
    
    if not patient_encounters.empty:
        most_recent_encounter = patient_encounters.iloc[0]
        most_recent_encounter_date = most_recent_encounter['START']
        most_recent_diagnosis = most_recent_encounter['DESCRIPTION']
        most_recent_reason = most_recent_encounter['REASONDESCRIPTION'] if pd.notna(most_recent_encounter['REASONDESCRIPTION']) else None
    
    # Find procedures related to most recent encounter
    recent_encounter_procedures = []
    if most_recent_encounter_date is not None and not patient_procedures.empty:
        # Get procedures within 24 hours of the most recent encounter
        recent_date = pd.to_datetime(most_recent_encounter_date)
        for _, proc in patient_procedures.iterrows():
            proc_date = pd.to_datetime(proc['START'])
            if abs((proc_date - recent_date).total_seconds()) < 86400:  # 24 hours in seconds
                recent_encounter_procedures.append({
                    'date': proc['START'],
                    'description': proc['DESCRIPTION'],
                    'reason': proc['REASONDESCRIPTION'] if pd.notna(proc['REASONDESCRIPTION']) else 'Not specified'
                })
    
    # Build enhanced patient profile
    profile = {
        'id': patient_id,
        'basic_info': {
            'first_name': patient['FIRST'],
            'last_name': patient['LAST'],
            'preferred_name': preferred_name if preferred_name else patient['FIRST'],
            'prefix': patient['PREFIX'] if 'PREFIX' in patient and pd.notna(patient['PREFIX']) else None,
            'suffix': patient['SUFFIX'] if 'SUFFIX' in patient and pd.notna(patient['SUFFIX']) else None,
            'age': age,
            'birthdate': patient['BIRTHDATE'],
            'gender': patient['GENDER'],
            'race': patient['RACE'],
            'ethnicity': patient['ETHNICITY'],
            'marital_status': patient['MARITAL'] if 'MARITAL' in patient and pd.notna(patient['MARITAL']) else 'Unknown',
            'birthplace': patient['BIRTHPLACE'] if 'BIRTHPLACE' in patient and pd.notna(patient['BIRTHPLACE']) else 'Unknown',
            'address': f"{patient['ADDRESS']}, {patient['CITY']}, {patient['STATE']} {patient['ZIP']}",
            'healthcare_expenses': patient['HEALTHCARE_EXPENSES'] if 'HEALTHCARE_EXPENSES' in patient else None,
            'healthcare_coverage': patient['HEALTHCARE_COVERAGE'] if 'HEALTHCARE_COVERAGE' in patient else None,
        },
        'conditions': [],
        'medications': [],
        'most_recent_encounter': {
            'date': most_recent_encounter_date,
            'description': most_recent_diagnosis,
            'reason': most_recent_reason,
            'procedures': recent_encounter_procedures
        },
        'encounters': [],
        'procedures': [],
        'allergies': patient_allergies['DESCRIPTION'].tolist() if not patient_allergies.empty else [],
        'social_determinants': []
    }
    
    # Find primary care provider from encounters
    for _, enc in patient_encounters.iterrows():
        if 'primary care' in str(enc['DESCRIPTION']).lower() or enc['ENCOUNTERCLASS'] == 'ambulatory':
            profile['basic_info']['primary_provider'] = enc['PROVIDER']
            profile['basic_info']['primary_organization'] = enc['ORGANIZATION']
            break
    
    # Process social determinants
    for _, social in patient_social.iterrows():
        social_info = {
            'category': social['CATEGORY'] if 'CATEGORY' in social else 'Unknown',
            'description': social['DESCRIPTION'] if 'DESCRIPTION' in social else 'Unknown'
        }
        profile['social_determinants'].append(social_info)
    
    # Process current conditions (ongoing conditions only)
    for _, condition in patient_conditions.iterrows():
        condition_info = {
            'date': condition['START'],
            'end_date': condition['STOP'] if pd.notna(condition['STOP']) else 'Ongoing',
            'description': condition['DESCRIPTION'],
            'code': condition['CODE']
        }
        if condition_info['end_date'] == 'Ongoing':
            profile['conditions'].append(condition_info)
    
    # Process current medications (ongoing medications only)
    for _, med in patient_medications.iterrows():
        med_info = {
            'date': med['START'],
            'end_date': med['STOP'] if pd.notna(med['STOP']) else 'Ongoing',
            'description': med['DESCRIPTION'],
            'reason': med['REASONDESCRIPTION'] if pd.notna(med['REASONDESCRIPTION']) else 'Not specified',
            'dispense': med['DISPENSES'] if 'DISPENSES' in med and pd.notna(med['DISPENSES']) else None,
            'dosage': med['DOSAGE'] if 'DOSAGE' in med and pd.notna(med['DOSAGE']) else None,
            'instructions': med['INSTRUCTIONS'] if 'INSTRUCTIONS' in med and pd.notna(med['INSTRUCTIONS']) else None
        }
        if med_info['end_date'] == 'Ongoing':
            profile['medications'].append(med_info)
    
    # Store only a few recent encounters for context
    recent_encounters_limit = 3
    counter = 0
    for _, enc in patient_encounters.iterrows():
        if counter >= recent_encounters_limit:
            break
        
        enc_info = {
            'date': enc['START'],
            'end_date': enc['STOP'],
            'type': enc['ENCOUNTERCLASS'],
            'description': enc['DESCRIPTION'],
            'reason': enc['REASONDESCRIPTION'] if pd.notna(enc['REASONDESCRIPTION']) else 'Not specified',
            'provider': enc['PROVIDER'],
            'organization': enc['ORGANIZATION']
        }
        profile['encounters'].append(enc_info)
        counter += 1
    
    return profile

# One Shot Prompt
normalsummary_output = '''# Discharge Summary for Sana377 Wolff180

**Patient Name:** Sana377 Wolff180  
**Preferred Name:** Sana377  
**Age:** 47  
**Gender:** Female  
**Visit Date:** November 4, 2021  
**Attending Physician:** Dr. Jane Doe, MD  
**Primary Care Provider:** Dr. John Smith, MD (Example Health Network)  
**Hospital:** Example General Hospital  
**Hospital Contact:** 800-555-1234  
**Primary Care Phone:** 800-555-9876  

---

## Reason for Visit
You visited Example General Hospital for a **scheduled procedure**, during which you received an **intramuscular injection**. This was not an emergency visit, and you did well during your time here.

---

## Summary of Your Health
You are managing several long-term conditions. These can sometimes affect how you feel or think, and it's okay to ask for help. Below is a list of your ongoing conditions for your reference:

- Type 2 Diabetes  
- High Blood Pressure (Hypertension)  
- High Triglycerides (Hypertriglyceridemia)  
- Anemia  
- Metabolic Syndrome  
- Prediabetes and Hyperglycemia  
- Severe Anxiety (Panic)  
- Past Miscarriage (First Trimester)  
- History of Unsatisfactory Housing

---

## What You Need to Do Next

### 1. **Follow-Up Care**
Please schedule a check-up with your **primary care provider (Dr. John Smith)** within **7 days** (by **November 11, 2021**). This is important to check on how you're feeling after the injection and review your ongoing care.

- **To schedule**: Call 800-555-9876  
- **Ask about**: Blood pressure and blood sugar follow-up, anxiety management options  

If transportation or housing is an issue for making your appointment, please let your provider know. Help is available.

---

### 2. **Medications**

Please continue taking your medications as directed. If you have any trouble remembering them, consider using a pill organizer or setting reminders on your phone.

| **Medication**                      | **Dosage**           | **When to Take**                 | **Purpose**            |
|------------------------------------|----------------------|----------------------------------|------------------------|
| Metformin (Extended Release)       | 500 mg               | 1 tablet **with breakfast**      | Controls blood sugar   |
| Hydrochlorothiazide                | 25 mg                | 1 tablet **in the morning**      | Lowers blood pressure  |
| Lisinopril                         | 10 mg                | 1 tablet **at night**            | Lowers blood pressure  |

No new medications were started during this visit.

---

### 3. **Important Warning Signs**

Please go to the **Emergency Room or call 911** if you experience:

- Trouble breathing  
- Severe chest pain or pressure  
- Sudden confusion or acting differently than usual  
- Fainting or feeling very dizzy  
- Very high or very low blood sugar (shaking, sweating, blurry vision)

---

## Mental Health Support

We know that managing anxiety can feel overwhelming. You're not alone. If you need support, here are some resources:

- **Mental Health Support Line**: 800-555-4321  
- **Crisis Line (24/7)**: 800-555-6789  

---

## Additional Information

- **Allergies:** None known  
- **Support Needs:** If housing or daily living is difficult, let your care team know. Help is available.  
- **Notes for Caregivers:** If you are helping Sana377, please ensure medication is taken as scheduled and watch for any signs of confusion or distress.

---

## Questions?

If you have any questions or are unsure about anything in this summary:

- **Hospital Main Line**: 800-555-1234  
- **Dr. John Smith's Office**: 800-555-9876  
- **Pharmacy Help**: Available through your health plan or local pharmacist  

---

**Thank you for trusting us with your care. You did great during your visit, and we're here to support you every step of the way.**
'''

deathsummary_output = '''# Discharge Summary for James L. Navarro

**Patient Name:** James L. Navarro  
**Age:** 72  
**Gender:** Male  
**Date of Admission:** March 24, 2025  
**Date of Passing:** March 27, 2025  
**Attending Physician:** Dr. Evelyn Moore, MD  
**Hospital:** Memorial Unity Hospital  
**Hospital Contact:** 800-555-1010  

---

## Summary of Hospital Stay

James was admitted to Memorial Unity Hospital on March 24, 2025, after experiencing shortness of breath, chest discomfort, and increasing fatigue. Upon evaluation, he was found to have **severe congestive heart failure**, likely due to long-standing **coronary artery disease** and a **recent, extensive myocardial infarction (heart attack)**.

Throughout his stay, the medical team worked closely to stabilize his condition. He was placed on continuous cardiac monitoring, received medications to support his heart and breathing, and was cared for with compassion and diligence.

Despite all efforts, James's condition continued to decline. On the evening of March 27, 2025, his heart rhythm became unstable and he went into cardiac arrest. Resuscitation efforts were initiated immediately but were ultimately unsuccessful. James passed away peacefully, surrounded by members of the care team.

---

## Cause of Death

**Primary Cause of Death:**  
- Acute myocardial infarction (heart attack)  
- Complicated by congestive heart failure and arrhythmia  

**Contributing Conditions:**  
- Hypertension (high blood pressure)  
- Type 2 Diabetes  
- Chronic kidney disease (Stage 3)

---

## Words to the Family

To James's family and loved ones,

We extend our deepest sympathies for your loss. James was treated with the utmost respect and dignity throughout his time with us. His condition was serious and, despite appropriate and timely medical intervention, the damage to his heart was too great to overcome.

Please know that he did not suffer in his final moments and was given every measure of comfort. His passing was peaceful.

If you have questions about his care or would like to speak with a member of the team, we welcome the opportunity to talk with you. We also encourage you to reach out for grief support or counseling resources—we are here to help during this difficult time.

---

## Support and Resources

| **Support Service**                     | **Contact Information**              |
|----------------------------------------|-------------------------------------|
| Hospital Bereavement Support           | 800-555-2020                        |
| Local Grief Counseling (Unity Care)    | 800-555-3030                        |
| Memorial Service Coordination          | 800-555-4040                        |
| Medical Records Request (if needed)    | 800-555-5050                        |

---

## Final Thoughts

James's warmth, courage, and character were evident in the short time we had with him. It was an honor to care for him. Please take the time you need to grieve and heal, and know that our thoughts are with you.

If you have questions about this summary or would like help understanding anything further, don't hesitate to contact us.

**With heartfelt sympathy,**  
**Dr. Evelyn Moore, MD and the Memorial Unity Care Team**
'''
childsummary_output = '''# Discharge Summary for Jordan Miles Thompson

**Patient Name:** Jordan Miles Thompson  
**Age:** 11  
**Gender:** Male  
**Date of Admission:** April 9, 2025  
**Date of Discharge:** April 12, 2025  
**Attending Physician:** Dr. Priya Nandini, MD  
**Hospital:** Oakridge Children's Medical Center  
**Contact:** (555) 764-8810  

---

## Reason for Admission

Jordan was admitted for evaluation and treatment of **moderate asthma exacerbation**. He was experiencing difficulty breathing, frequent coughing, and wheezing that did not improve with his home inhaler (albuterol). These symptoms were triggered by an upper respiratory infection.

---

## Hospital Course

Upon arrival, Jordan was treated with **nebulized bronchodilators**, **oral steroids**, and **oxygen therapy** to help him breathe more easily. His symptoms improved over the first 48 hours, and he remained alert, active, and cooperative during his hospital stay.

He was monitored for any signs of fatigue, oxygen desaturation, or secondary infection. By April 11, Jordan no longer required oxygen and was breathing comfortably on his own. He tolerated food well, maintained hydration, and resumed regular sleep and activity patterns.

---

## Medications

| **Medication**            | **Dosage**             | **Instructions**                             | **Duration**         |
|---------------------------|------------------------|----------------------------------------------|-----------------------|
| Albuterol (inhaler)       | 2 puffs                | Every 4-6 hours as needed for wheezing       | As Need               |
| Prednisolone (oral)       | 15 mg once daily       | Take in the morning with food                | 5 days total         |
| Fluticasone (inhaled)     | 1 puff twice daily     | Use with spacer; rinse mouth after use       | Ongoing (maintenance)|

---

## Discharge Plan and Instructions

To Jordan and his family,

Jordan has responded well to treatment, and we are happy to see him breathing easier and back to his normal self. At this point, it is safe for him to continue his recovery at home. Please follow the medication schedule provided, and continue using his **maintenance inhaler daily**, even if he feels fine — this helps prevent future flare-ups.

We also reviewed with you and Jordan how to properly use his inhalers and spacer. If you're unsure about the technique at any point, don't hesitate to ask your pharmacy or pediatrician to review it again.

---

## When to Call the Doctor or Return to the ER

Please contact your pediatrician or return to the hospital if you notice any of the following:
- Increased difficulty breathing
- Persistent wheezing that does not improve after using albuterol
- Trouble speaking in full sentences
- Bluish lips or fingernails
- High fever (above 102.5°F)

---

## Follow-Up

| **Provider**                 | **Purpose**               | **When**             | **Contact**        |
|-----------------------------|---------------------------|----------------------|--------------------|
| Dr. Rachel Sung (Pediatrics)| Asthma follow-up & refill | Within 3-5 days      | (555) 764-5520     |
| Pulmonology (if needed)     | Referral for asthma mgmt  | TBD if symptoms persist | Through PCP        |

---

## Notes for Jordan

Jordan — we're really proud of how brave you were during your stay! You asked great questions and took your breathing treatments like a champ. Keep up the good work with your inhaler, and make sure to let a grown-up know if you start feeling tight in your chest again.

---

**Prepared by:**  
Dr. Priya Nandini, MD  
Oakridge Children's Medical Center  
April 12, 2025
'''

def format_medical_history_prompt(patient_history, target_audience):
    """
    Format the patient history into a personalized prompt for the LLM,
    focusing on the most recent visit and only including sections with actual data.
    """
    # Define prompt based on target audience
    if target_audience == "pediatric":
        audience_instruction = f"""The patient is a minor (age {patient_history['basic_info']['age']}). 
Create a simplified discharge summary focused on their most recent hospital visit that would be appropriate for a child or adolescent patient and their parents/guardians.
Use warm, reassuring language and simple explanations of medical concepts. 
If the child is very young (under 10), include elements that make the summary more engaging for kids.
Address both the child and the parent/guardian in the summary."""
    elif target_audience == "elderly":
        audience_instruction = f"""The patient is elderly (age {patient_history['basic_info']['age']}). 
Create a simplified discharge summary focused on their most recent hospital visit with visual indicators for important sections, clear medication instructions, 
and straightforward follow-up directions. Be respectful and avoid condescending language.
Remember that elderly patients often have more complex health situations and may be taking multiple medications."""
    elif target_audience == "cognitive_impairment":
        audience_instruction = """The patient has a condition that may affect cognition (e.g., dementia, intellectual disability, brain injury). 
Create a very simplified discharge summary focused on their most recent hospital visit with visual cues, simple language, step-by-step instructions, 
and information directed at both the patient and their caregiver(s).
Use short sentences, simple words, and clear organization."""
    elif target_audience == "esl":
        audience_instruction = """Create a discharge summary for someone with limited English proficiency. 
Focus on their most recent hospital visit using simple vocabulary, avoid idioms and medical jargon, and include visual indicators where helpful.
Use short, clear sentences and provide explanations for necessary medical terms."""
    else:
        audience_instruction = """Create a standard discharge summary in simple language that most patients can understand.
Focus on their most recent hospital visit, balancing medical accuracy with clear explanations."""
    
    # Format patient demographics with more personalized information
    basic_info = patient_history['basic_info']
    
    # Build formal name with prefix/suffix if available
    full_name = ""
    if basic_info['prefix']:
        full_name += f"{basic_info['prefix']} "
    full_name += f"{basic_info['first_name']} {basic_info['last_name']}"
    if basic_info['suffix']:
        full_name += f" {basic_info['suffix']}"
    
    preferred_name = basic_info['preferred_name'] if basic_info['preferred_name'] else basic_info['first_name']
    
    demographics = f"""
Patient Information:
- Full Name: {full_name}
- Preferred Name: {preferred_name}
- Age: {basic_info['age']}
- Gender: {basic_info['gender']}
- Race: {basic_info['race']}
- Ethnicity: {basic_info['ethnicity']}
"""
    
    # Only add other demographic fields if they're meaningful
    if basic_info['marital_status'] and basic_info['marital_status'] != 'Unknown':
        demographics += f"- Marital Status: {basic_info['marital_status']}\n"
    
    if basic_info['birthplace'] and basic_info['birthplace'] != 'Unknown':
        demographics += f"- Birthplace: {basic_info['birthplace']}\n"
    
    if basic_info['address']:
        demographics += f"- Address: {basic_info['address']}\n"
    
    # Add primary care provider info if available
    if 'primary_provider' in basic_info and basic_info['primary_provider']:
        demographics += f"- Primary Care Provider: {basic_info['primary_provider']}\n"
        if 'primary_organization' in basic_info and basic_info['primary_organization']:
            demographics += f"- Primary Care Organization: {basic_info['primary_organization']}\n"
    
    # Add social determinants if available and relevant
    social_info = ""
    relevant_social_categories = ['education', 'housing', 'employment', 'food', 'transportation', 'caregiving']
    relevant_social_determinants = []
    
    for social in patient_history['social_determinants']:
        category = social['category'].lower() if isinstance(social['category'], str) else ""
        if any(rel_cat in category for rel_cat in relevant_social_categories):
            relevant_social_determinants.append(social)
    
    if relevant_social_determinants:
        social_info = "\nSocial Context:\n"
        for social in relevant_social_determinants:
            social_info += f"- {social['category']}: {social['description']}\n"
    
    # Information about most recent encounter (the one they're being discharged from)
    discharge_info = ""
    if patient_history['most_recent_encounter']['date']:
        discharge_info = "\nCurrent Discharge Information:\n"
        discharge_info += f"- Visit Date: {patient_history['most_recent_encounter']['date']}\n"
        
        if patient_history['most_recent_encounter']['description']:
            discharge_info += f"- Reason for Visit: {patient_history['most_recent_encounter']['description']}\n"
        
        if patient_history['most_recent_encounter']['reason']:
            discharge_info += f"- Specific Condition: {patient_history['most_recent_encounter']['reason']}\n"
        
        # Add procedures performed during this visit
        if patient_history['most_recent_encounter']['procedures']:
            discharge_info += "- Procedures Performed:\n"
            for proc in patient_history['most_recent_encounter']['procedures']:
                discharge_info += f"  * {proc['description']}\n"
    
    # Format current conditions - only if there are any
    conditions_text = ""
    if patient_history['conditions']:
        conditions_text = "\nCurrent Health Conditions:\n"
        for cond in patient_history['conditions']:
            # Skip conditions that might be social determinants rather than medical
            condition_desc = cond['description'].lower()
            skip_terms = ['employment', 'education', 'refugee', 'victim', 'criminal record', 'social']
            if any(term in condition_desc for term in skip_terms):
                continue
            
            conditions_text += f"- {cond['description']} (since {cond['date']})\n"
    
    # Format current medications - only if there are any
    medications_text = ""
    if patient_history['medications']:
        medications_text = "\nCurrent Medications:\n"
        for med in patient_history['medications']:
            med_text = f"- {med['description']} (Reason: {med['reason']})"
            if med['dosage']:
                med_text += f", Dosage: {med['dosage']}"
            if med['instructions']:
                med_text += f", Instructions: {med['instructions']}"
            medications_text += med_text + "\n"
    
    # Format allergies - only if there are any
    allergies_text = ""
    if patient_history['allergies']:
        allergies_text = "\nAllergies:\n"
        for allergy in patient_history['allergies']:
            allergies_text += f"- {allergy}\n"
    else:
        allergies_text = "\nAllergies:\n- No known allergies\n"

    # Fake Hospital Name and Provider Info Text
    hospital_info_text = '''
Provider information:
- Hospital Name: Example General Hospital
- Hospital Address: 1234 Health St, Cityville, ST 12345
- Hospital Phone: 800-555-1234
- Attending Physician: Dr. Jane Doe, MD
- Attending Physician Phone: 800-555-5678
- Primary Care Provider: Dr. John Smith, MD
- Primary Care Provider Phone: 800-555-9876
- Primary Care Provider Organization: Example Health Network
- Crisis Line: 800-555-6789
- Mental Health Support: 800-555-4321
- Substance Use Support: 800-555-8765
    '''


    # Create the full prompt with more personalization instructions
    prompt = f"""
You are a compassionate and detail-oriented medical communication specialist who crafts personalized, empathetic, and easy-to-understand discharge summaries. Your summaries should make the patient (or their family, in case of death) feel seen, respected, and well-informed about their care.  

{audience_instruction}

PATIENT INFORMATION:
{demographics}
{social_info}
{discharge_info}
{conditions_text}
{medications_text}
{allergies_text}
{hospital_info_text}

Create a personalized discharge summary that:
1. Addresses the patient by their preferred name ({preferred_name}) and makes them feel seen as a person, not just a medical case
2. Focuses specifically on their most recent visit and what they were treated for
3. Clearly explains any follow-up care needed, including when they should return for a check-up
4. Includes specific medications they need to take (with clear schedule and dosage)
5. Includes warning signs that should prompt them to seek immediate care
6. Uses appropriate visual formatting for the target audience

Write in a warm, empathetic tone that acknowledges the patient's individual circumstances. Avoid generic language and placeholder text like [insert medication name] or [insert time frame]. If specific information is missing, make appropriate recommendations based on the condition.

Additionally, there are a few edge cases you should watch out for:
- If the patient has a history of substance abuse, be sensitive to that in your language.
- If the patient has a history of mental health issues, be careful to avoid stigmatizing language.
- If the patient has a history of trauma, be mindful of that in your language and avoid triggering terms.
- If the patient has a history of non-compliance with medications or follow-up care, be gentle but firm in your recommendations.
- If the patient has a history of complex medical conditions, be sure to explain things clearly and simply.
- If the patient has died, or is receiving a death certificate, instead of creating a summary for the patient create a summary that is appropriate for the family or next of kin to explain why their loved one has died and what the next steps are. 
- If the patient is a minor, be sure to include information that is appropriate for both the child and their parents/guardians, with a respect for asking a parent to administer medications and follow-up care.
- If the patient is elderly, be sure to include information that is appropriate for their age group and any potential cognitive impairments they may have, as well as any social determinants of health that may affect their care.
Make the patient feel that this summary was created specifically for them. Use real details from their record throughout the summary.

Important discharge information should be clearly highlighted. Medication instructions should be particularly clear and easy to follow. 

### **Patient-Specific Considerations:**  
Before writing, carefully review:  
- **Demographics** (age, preferred name, language needs)  
- **Medical history** (conditions, medications, allergies)  
- **Social determinants of health** (housing, support system, financial barriers)  
- **Special circumstances** (mental health, trauma history, substance use, cognitive impairments, pediatric/geriatric needs)  

### **Key Requirements for the Discharge Summary:**  
1. **Personalization**  
   - Address the patient by their **preferred name** ({preferred_name}).  
   - Acknowledge their **individual situation** (e.g., "We know managing diabetes can be challenging, but you're doing great by staying on top of your care.").  
   - Avoid **generic templates**—every sentence should reflect their real medical details.  

2. **Clear Focus on This Visit**  
   - Explain **why they were admitted** and **what treatment they received**.  
   - Keep it relevant—don't overload with unrelated past medical history.  

3. **Follow-Up Care Instructions**  
   - Specify:  
     - **When to return** (e.g., "Please see Dr. Smith in 7 days, by [exact date]").  
     - **Who to contact** if issues arise (e.g., "Call Cardiology at [phone] if you have chest pain.").  
   - If the patient has a history of **missed appointments**, gently emphasize importance (e.g., "We know life gets busy, but this follow-up is key to your recovery.").  

4. **Medication Instructions (Critical!)**  
   - List **each medication**, **dosage**, **schedule**, and **purpose** in a **bullet-point format**.  
     - Example:  
       - **Metformin 500mg**: Take **1 pill with breakfast and dinner** to help control blood sugar.  
   - Highlight **new medications** or **changes** from prior prescriptions.  
   - If the patient has **history of non-compliance**, add a supportive note (e.g., "We recommend setting a daily alarm to help remember your pills."). 
   - Ensure special medications like Antibiotics have proper warnings such as to not take if pregnant or breastfeeding, and to take with food.
   - Include **specific instructions** for medications (e.g., "Take with food" or "Do not crush")
   - For some medications, like antibiotics, include a warning about potential side effects, such as "If you develop a rash or have difficulty breathing, stop taking this medication and call us immediately.", and instructions to finish the course with a warning about bacterial resistance.

5. **Warning Signs (Red Flags)**  
   - List **symptoms that require urgent care** (e.g., "Seek ER care if you have:  
     - Shortness of breath  
     - Sudden confusion").  

6. **Formatting for Clarity**  
   - Use **bold**, bullet points, and spacing to improve readability.  
   - For **low-literacy audiences**, simplify language (avoid jargon like "myocardial infarction"—say "heart attack").  
   - For **minors**, include **parent-friendly instructions** (e.g., "Give your child 5mL of amoxicillin every 12 hours.").  

### **Special Edge Cases & Sensitivity:**  
- **Substance use history**: Avoid judgmental terms (e.g., say "managed withdrawal" not "drug abuse").  
- **Mental health/Trauma**: Use neutral language (e.g., "We recommend counseling for support" vs. "You need therapy.").  
- **Deceased patients**: Write for the **family**, explaining cause of death and next steps (e.g., funeral home contact).  
- **Elderly patients**: Include reminders for **memory aids** (e.g., pill organizers) and **social support needs**.  

### **Final Checks:**  
- **No placeholders** (e.g., [insert date] → use real data).  
- **Warm tone** (e.g., "We're here to support you.").  
- **Highlight critical info** (e.g., **"!Call 911 if you have trouble breathing."**).  
"""
    
    return prompt


def create_deepseek_thinking_prompt(original_prompt):
    """
    Create a prompt for Deepseek-R1 to analyze the discharge summary requirements
    and think through how to create the best possible summary.
    """
    thinking_prompt = f"""
You are a medical professional with expertise in creating clear and personalized discharge summaries.
You need to carefully think through how to create the best discharge summary for a specific patient.
I'll provide you with the patient's information and your task is to:

1. Analyze the key elements listed below that need special attention
2. Ask yourself questions about what specific information is needed to provide the patient the most wholistic care, and make presumptions about what likely happened in the discharge case. Come up with a story in your head of how the patient got to this point, and what they need to know to be successful in their care
3. Examine the patient's specific circumstances and how they affect the discharge summary
4. Consider the appropriate tone, language, and structure for this specific patient, especially when dealing with mental health disorders. If a patient has a history of depression, consider asking them questions that would help them feel more comfortable and supported.
5. Think about potential pitfalls to avoid in your tone, the information you include, and the focus you have
6. Estimate the follow up actions that would be appropriate from a physician, such as if a followup is necessary, the length of time appropriate to wait until the next followup and the likely reason for a followup if needed
7. Think about the medications the patient is on, what they are for and when they are from (predict if they are still taking them), and how the medications they are currently taking might interact with each other
8. Consider the patient's social determinants of health and how they might affect their care, if a patient has a history of being unhoused or having a history of substance use, how might that affect their care and the advice a physician would give them
9. Think about the patient's history of non-compliance with medications or follow-up care, and come up with a way to explain to the patient why this is important and how they will see long term benefits even with short term discomfort
10. If this is a special case, such as a minor or an elderly patient, someone who has died, someone who is unresponsive, or someone who has a cognitive impairment, think about how that would affect the discharge summary and what you would need to include in the summary to make it appropriate for that patient, their caregivers, or their surviving kin
11. Utilize the hospital name and provider information anywhere you require including it

Here is the patient information:

{original_prompt}

Now, think step by step about how you would create this discharge summary. Ask yourself questions, consider different approaches, and really explore what would make this summary most effective and personalized for this specific patient. You will then generate a prompt which focuses on the data you identified as important to ensure the next step of the process is as effective as possible.
Your task is to create a prompt that will be sent to the Llama4 model. This prompt should include all the information you have thought about, and should be clear and easy to understand. The goal is to create a discharge summary that is personalized, empathetic, and medically accurate.
"""
    return thinking_prompt

def call_deepseek_for_thinking(thinking_prompt):
    """
    Call Deepseek-R1 model to analyze the patient information and generate a thinking process.
    """
    print("\n" + "="*80)
    print("🧠 STEP 2: CALLING DEEPSEEK-R1 FOR THINKING PROCESS")
    print("="*80)
    print("Sending prompt to Deepseek-R1 for analysis...")
    
    try:
        response = requests.post(
            "http://10.20.31.1:11434/api/chat",
            json={
                "model": "deepseek-r1",
                "messages": [{"role": "user", "content": thinking_prompt}],
                "stream": False
            }
        )
        
        deepseek_thinking = response.json()["message"]["content"]
        print("\n📝 DEEPSEEK-R1 THINKING ANALYSIS:")
        print("-"*80)
        print(deepseek_thinking)
        print("-"*80)
        
        return deepseek_thinking
    except Exception as e:
        print(f"Error calling Deepseek-R1: {str(e)}")
        return "Error: Unable to complete the thinking process with Deepseek-R1."

def create_llama_final_prompt(original_prompt, deepseek_thinking, patient_history):
    """
    Create a prompt for Llama4 that includes both the original prompt, Deepseek's thinking,
    and the appropriate one-shot example based on patient characteristics.
    """
    # Determine which example to use based on patient characteristics
    patient_age = patient_history['basic_info']['age']
    
    # Check if the patient has died - look through conditions and encounters for death indicators
    is_deceased = False
    death_keywords = ['death', 'deceased', 'mortality', 'died', 'expire', 'fatal']
    
    # Check encounters for death indicators
    for encounter in patient_history['encounters']:
        desc = str(encounter['description']).lower()
        if any(keyword in desc for keyword in death_keywords):
            is_deceased = True
            break
    
    # Check conditions for death indicators if not already found
    if not is_deceased:
        for condition in patient_history['conditions']:
            desc = str(condition['description']).lower()
            if any(keyword in desc for keyword in death_keywords):
                is_deceased = True
                break
                
    # Select appropriate example
    if is_deceased:
        example_output = deathsummary_output
        print("Using death summary template - patient appears to be deceased")
    elif patient_age < 18:
        example_output = childsummary_output
        print(f"Using pediatric summary template - patient is {patient_age} years old")
    else:
        example_output = normalsummary_output
        print(f"Using standard summary template - patient is {patient_age} years old adult")

    llama_prompt = f"""
    You are a compassionate, detail-oriented medical communication specialist tasked with creating a highly personalized and effective discharge summary.

    I'm providing you with three key resources:

    1. ORIGINAL PATIENT INFORMATION AND REQUIREMENTS:
    {original_prompt}

    2. EXPERT ANALYSIS AND CONSIDERATIONS:
    This is analysis by deepseek R1, which has thought through the key elements of the discharge summary and how to best personalize it for this specific patient:
    {deepseek_thinking}

    Your task is to synthesize this information and create a discharge summary that is:
    - Perfectly tailored to this specific patient's needs and circumstances
    - Clear, warm, and empathetic in tone
    - Medically accurate while being easy to understand
    - Well-formatted for the target audience
    - Discuss all medications, when they should be taken, their dosages, and what they do / why the patient needs them
    - Includes all medically relevant details which were discussed in the patients chart
    - Gives the patient actionable next steps to further improve their health
    - Addresses all the key requirements and considerations from both sources
    - Format the summary in markdown, using headings, bullet points, and other formatting to make it easy to read

    Please create the final discharge summary that would be given to the patient or their caregiver. Remember to list each section only once, and highlight important information clearly. Use the format most appropriate for the hospital discharge setting.

    3. EXAMPLE OUTPUT:
    Here is an example output as a reference for formatting only, you should deviate from this reference if it is not appropriate to the clinical needs being addressed. Remember, this is just a reference for formatting, and you should use it as a template. Do not reference or use any clinical data or items in this example. Replace the items with information found under 1. ORIGINAL PATIENT INFORMATION AND REQUIREMENTS:, such as patient history, medications, and any other relevant information.
    {example_output}
    """
    return llama_prompt

def call_llama_for_final_output(llama_prompt):
    """
    Call Llama4 model to generate the final discharge summary.
    """
    print("\n" + "="*80)
    print("📋 STEP 3: CALLING LLAMA4 FOR FINAL DISCHARGE SUMMARY")
    print("="*80)
    print("Sending enhanced prompt to Llama4...")
    
    try:
        response = requests.post(
            "http://10.20.31.1:11434/api/chat",
            json={
                "model": "aravhawk/llama4",
                "messages": [{"role": "user", "content": llama_prompt}],
                "stream": False
            }
        )
        
        final_summary = response.json()["message"]["content"]
        print("\n🏥 FINAL DISCHARGE SUMMARY FROM LLAMA4:")
        print("-"*80)
        print(final_summary)
        print("-"*80)
        
        return final_summary
    except Exception as e:
        print(f"Error calling Llama4: {str(e)}")
        return "Error: Unable to generate the final discharge summary with Llama4."

def generate_multi_model_discharge_summary(patient_id, target_audience="standard"):
    """
    Generate a personalized discharge summary using a multi-model approach:
    1. Generate the initial prompt
    2. Use Deepseek-R1 for thinking and analysis
    3. Use Llama4 for the final high-quality output
    """
    print("\n" + "="*80)
    print(f"🔄 STARTING MULTI-MODEL DISCHARGE SUMMARY GENERATION FOR PATIENT {patient_id}")
    print("="*80)
    
    # STEP 1: Get patient history and format the initial prompt
    print("\n" + "="*80)
    print("📊 STEP 1: GENERATING INITIAL PROMPT")
    print("="*80)
    
    patient_history = get_complete_patient_history(patient_id)
    original_prompt = format_medical_history_prompt(patient_history, target_audience)
    
    print(f"\nPatient: {patient_history['basic_info']['first_name']} {patient_history['basic_info']['last_name']} (Age: {patient_history['basic_info']['age']})")
    if patient_history['most_recent_encounter']['date']:
        print(f"Most recent visit: {patient_history['most_recent_encounter']['date']}")
        print(f"Reason: {patient_history['most_recent_encounter']['description']}")
        if patient_history['most_recent_encounter']['reason']:
            print(f"Specific condition: {patient_history['most_recent_encounter']['reason']}")
    
    print("\n📝 INITIAL PROMPT:")
    print("-"*80)
    print(original_prompt)
    print("-"*80)
    
    # STEP 2: Create thinking prompt and call Deepseek-R1
    thinking_prompt = create_deepseek_thinking_prompt(original_prompt)
    deepseek_thinking = call_deepseek_for_thinking(thinking_prompt)
    
    # STEP 3: Create final prompt and call Llama4
    llama_prompt = create_llama_final_prompt(original_prompt, deepseek_thinking, patient_history)
    print("\n📝 FINAL ENHANCED PROMPT FOR LLAMA4:")
    print("-"*80)
    print(llama_prompt[:500] + "... [truncated for display]")
    print("-"*80)
    
    final_summary = call_llama_for_final_output(llama_prompt)
    
    print("\n" + "="*80)
    print("✅ MULTI-MODEL DISCHARGE SUMMARY GENERATION COMPLETE")
    print("="*80)
    
    return {
        "patient_id": patient_id,
        "target_audience": target_audience,
        "patient_name": f"{patient_history['basic_info']['first_name']} {patient_history['basic_info']['last_name']}",
        "patient_age": patient_history['basic_info']['age'],
        "most_recent_visit": patient_history['most_recent_encounter'],
        "original_prompt": original_prompt,
        "deepseek_thinking": deepseek_thinking,
        "final_prompt": llama_prompt,
        "final_summary": final_summary
    }

# Function to estimate tokens
def estimate_tokens(text):
    try:
        import tiktoken
        encoding = tiktoken.get_encoding("cl100k_base")
        num_tokens = len(encoding.encode(text))
        return num_tokens
    except:
        # Rough estimation if tiktoken not available
        return len(text.split()) * 1.3

# Run tests for each patient category
def run_multi_model_category_tests():
    print("\n" + "="*80)
    print("🚀 RUNNING MULTI-MODEL TESTS FOR EACH PATIENT CATEGORY")
    print("="*80)
    
    results = []
    
    # 1. Pediatric patient
    random_pediatric = minors['Id'].sample(1).iloc[0]
    print(f"\n🧒 Selected pediatric patient ID: {random_pediatric}")
    pediatric_result = generate_multi_model_discharge_summary(random_pediatric, "pediatric")
    if pediatric_result:
        display(Markdown(pediatric_result["final_summary"]))
        results.append(pediatric_result)
    
    # 2. Elderly patient
    random_elderly = elderly['Id'].sample(1).iloc[0]
    print(f"\n👴 Selected elderly patient ID: {random_elderly}")
    elderly_result = generate_multi_model_discharge_summary(random_elderly, "elderly")
    if elderly_result:
        display(Markdown(elderly_result["final_summary"]))
        results.append(elderly_result)
    
    # 3. Patient with potential cognitive impairment (or another elderly if none found)
    if 'all_special_needs_patients' in globals() and len(all_special_needs_patients) > 0:
        random_special_needs = random.choice(list(all_special_needs_patients))
        print(f"\n🧠 Selected cognitive impairment patient ID: {random_special_needs}")
        special_needs_result = generate_multi_model_discharge_summary(random_special_needs, "cognitive_impairment")
        if special_needs_result:
            display(Markdown(special_needs_result["final_summary"]))
            results.append(special_needs_result)
    else:
        print("\n⚠️ No patients with cognitive impairment found. Using another elderly patient.")
        random_special_needs = elderly['Id'].sample(1).iloc[0]
        print(f"🧠 Selected alternative elderly patient ID: {random_special_needs}")
        special_needs_result = generate_multi_model_discharge_summary(random_special_needs, "cognitive_impairment")
        if special_needs_result:
            display(Markdown(special_needs_result["final_summary"]))
            results.append(special_needs_result)
    
    # Display comparative results and metrics
    print("\n" + "="*80)
    print("📊 COMPARATIVE RESULTS AND METRICS")
    print("="*80)
    
    for result in results:
        print(f"\n----- {result['target_audience'].upper()} DISCHARGE SUMMARY FOR {result['patient_name']} (Age: {result['patient_age']}) -----")
        
        # Calculate and display token metrics
        original_tokens = estimate_tokens(result["original_prompt"])
        thinking_tokens = estimate_tokens(result["deepseek_thinking"])
        final_prompt_tokens = estimate_tokens(result["final_prompt"])
        summary_tokens = estimate_tokens(result["final_summary"])
        
        print(f"Original prompt tokens: ~{original_tokens}")
        print(f"Deepseek thinking tokens: ~{thinking_tokens}")
        print(f"Final enhanced prompt tokens: ~{final_prompt_tokens}")
        print(f"Final summary tokens: ~{summary_tokens}")
        
        # Display a brief summary of what Deepseek thought about
        deepseek_summary = result["deepseek_thinking"][:300] + "..." if len(result["deepseek_thinking"]) > 300 else result["deepseek_thinking"]
        print(f"\nDeepseek thinking summary: {deepseek_summary}")
        
        # Display the final summary
        print("\n📋 FINAL DISCHARGE SUMMARY:")
        display(Markdown(result["final_summary"]))
        print("\n" + "-"*80)
    
    return results

# Function to compare single model vs multi-model output
def compare_approaches(patient_id, target_audience="standard"):
    """
    Generate discharge summaries using both approaches and compare them.
    """
    print("\n" + "="*80)
    print(f"🔍 COMPARING SINGLE-MODEL VS MULTI-MODEL APPROACH FOR PATIENT {patient_id}")
    print("="*80)

    # Get patient history first so we can use it for both approaches
    patient_history = get_complete_patient_history(patient_id)
    
    # Get single-model output (direct Llama4)
    print("\n📋 GENERATING SINGLE-MODEL OUTPUT (LLAMA4 ONLY)")
    single_model_result = generate_simplified_discharge_summary(patient_id, target_audience)
    
    # Get multi-model output
    print("\n📋 GENERATING MULTI-MODEL OUTPUT (DEEPSEEK-R1 + LLAMA4)")
    multi_model_result = generate_multi_model_discharge_summary(patient_id, target_audience)
    
    # Display comparison
    print("\n" + "="*80)
    print("📊 COMPARISON RESULTS")
    print("="*80)
    
    print(f"\nPatient: {multi_model_result['patient_name']} (Age: {multi_model_result['patient_age']})")
    print(f"Target audience: {target_audience}")
    
    print("\n🔹 SINGLE-MODEL SUMMARY (LLAMA4 ONLY):")
    display(Markdown(single_model_result["simplified_summary"]))
    
    print("\n🔹 MULTI-MODEL SUMMARY (DEEPSEEK-R1 + LLAMA4):")
    display(Markdown(multi_model_result["final_summary"]))
    
    # Calculate some basic metrics
    single_model_tokens = estimate_tokens(single_model_result["simplified_summary"])
    multi_model_tokens = estimate_tokens(multi_model_result["final_summary"])
    
    print("\n📊 BASIC METRICS:")
    print(f"Single-model summary length: ~{single_model_tokens} tokens")
    print(f"Multi-model summary length: ~{multi_model_tokens} tokens")
    
    return {
        "patient_id": patient_id,
        "target_audience": target_audience,
        "single_model_result": single_model_result,
        "multi_model_result": multi_model_result
    }

# Run a side-by-side comparison test
def run_comparison_test():
    print("\n" + "="*80)
    print("🔄 RUNNING SIDE-BY-SIDE COMPARISON TEST")
    print("="*80)
    
    # Select a random elderly patient
    random_elderly = elderly['Id'].sample(1).iloc[0]
    print(f"\nSelected elderly patient ID: {random_elderly}")
    
    # Run comparison
    comparison_result = compare_approaches(random_elderly, "elderly")
    
    return comparison_result

# Entry point function
def main():
    """
    Main entry point for the multi-model discharge summary generation system.
    """
    print("\n" + "="*80)
    print("🏥 MULTI-MODEL DISCHARGE SUMMARY GENERATION SYSTEM")
    print("="*80)
    print("This system uses a three-step approach:")
    print("1. Generate prompts using patient data")
    print("2. Call Deepseek-R1 for thinking and analysis")
    print("3. Call Llama4 for the final high-quality discharge summary")
    
    # Load necessary data (these would be defined elsewhere in your notebook)
    print("\nEnsure that these dataframes are defined in your environment:")
    print("- patients_df")
    print("- conditions_df")
    print("- medications_df")
    print("- encounters_df")
    print("- minors (subset of patients_df where age < 18)")
    print("- elderly (subset of patients_df where age >= 65)")
    print("- all_special_needs_patients (set of patient IDs with cognitive conditions)")
    
    # Run tests
    print("\nChoose one of the following tests to run:")
    print("1. run_multi_model_category_tests() - Test with pediatric, elderly, and cognitive impairment patients")
    print("2. compare_approaches(patient_id, target_audience) - Compare single-model vs multi-model approaches")
    print("3. run_comparison_test() - Run a side-by-side comparison with a random elderly patient")
    
    return "System initialized and ready to run tests."

# Execute main function if run directly
if __name__ == "__main__":
    main()
    print("Running multi-model discharge summary generation system...")
    # Run the tests and store results
    category_results = run_multi_model_category_tests()
    print("Running comparison test...")
    comparison_result = run_comparison_test()
    print("All tests completed.")

    # Save category test results to a file
    with open("discharge_summary_category_tests.txt", "w") as f:
        f.write("="*80 + "\n")
        f.write("MULTI-MODEL CATEGORY TEST RESULTS\n")
        f.write("="*80 + "\n\n")
        
        for result in category_results:
            f.write(f"TARGET AUDIENCE: {result['target_audience'].upper()}\n")
            f.write(f"PATIENT: {result['patient_name']} (Age: {result['patient_age']})\n")
            f.write("-"*80 + "\n")
            f.write(result['final_summary'])
            f.write("\n\n" + "="*80 + "\n\n")
        
        print(f"Category test results saved to discharge_summary_category_tests.txt")

    # Save single-model comparison results to a file
    with open("discharge_summary_single_model.txt", "w") as f:
        f.write("="*80 + "\n")
        f.write("SINGLE-MODEL RESULTS (LLAMA4 ONLY)\n")
        f.write("="*80 + "\n\n")
        
        f.write(f"PATIENT: {comparison_result['multi_model_result']['patient_name']} (Age: {comparison_result['multi_model_result']['patient_age']})\n")
        f.write(f"TARGET AUDIENCE: {comparison_result['target_audience']}\n\n")
        
        f.write("-"*80 + "\n")
        f.write(comparison_result['single_model_result']['simplified_summary'])
        
        print(f"Single-model results saved to discharge_summary_single_model.txt")

    # Save multi-model comparison results to a file  
    with open("discharge_summary_multi_model.txt", "w") as f:
        f.write("="*80 + "\n")
        f.write("MULTI-MODEL RESULTS (DEEPSEEK-R1 + LLAMA4)\n")
        f.write("="*80 + "\n\n")
        
        f.write(f"PATIENT: {comparison_result['multi_model_result']['patient_name']} (Age: {comparison_result['multi_model_result']['patient_age']})\n")
        f.write(f"TARGET AUDIENCE: {comparison_result['target_audience']}\n\n")
        
        f.write("-"*80 + "\n")
        f.write(comparison_result['multi_model_result']['final_summary'])
        
        print(f"Multi-model results saved to discharge_summary_multi_model.txt")
```
